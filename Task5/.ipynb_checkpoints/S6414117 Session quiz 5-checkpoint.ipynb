```python
import numpy as np

name= str(input("Give the name of the file:"))
x, y = np.loadtxt(name, unpack=True)
print(f"x=", x)
print(f"y=", y)

x_max=np.max(x)
x_min=np.min(x)
y_max=np.max(y)
y_min=np.min(y)
x_mean=np.mean(x)
y_mean=np.mean(y)

print(f"for x, maximum= {x_max}, minimum= {x_min}, mean= {x_mean}")
print(f"for y, maximum= {y_max}, minimum= {y_min}, mean= {y_mean}")

y_med=np.median(y)
y2 = (y - y_med)**2
print(f"y2 array= {y2}")
```

```python
import matplotlib.pyplot as plt
import numpy as np


x = np.loadtxt('ngc6946.txt', unpack=True)


x= x.reshape(40, 50, 70)
print (f"the right shape= {x}")

z_middle= x[:, 25, 35]
print(f"z profile in the middle= {z_middle}")

z_max=z_middle.max()
z_min=z_middle.min()
z_sum=z_middle.sum()
z_mean=z_middle.mean()
z_std=z_middle.std()

print(f"for the z-profile: maximum= {z_max}, minimum= {z_min}, sum= {z_sum}, mean= {z_mean}, standard deviation= {z_std}")

plt.plot(z_middle)
plt.text(1, z_max, f"max={z_max}, min={z_min}, mean={z_mean}, standard deviation={z_std}")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.loadtxt('ngc6946.txt', unpack=True)


x= x.reshape(40, 50, 70)

plane=x[0,:,:]
mean=np.mean(plane)
std=np.std(plane)

levels=[mean, mean + std, mean + 2*std]

plt.imshow(plane, origin='lower', cmap='jet')
plt.colorbar()

plt.contour(plane, levels=levels, colors='white')

plt.show()
```

```python
class Galaxy:    
    """Class for redshift calculation of galaxies"""
        
    caH_0 = 3968.5      # Calcium H line in rest frame
    c=3.0e5
    def __init__(self, name, calciumHline=caH_0):
        self.name = name
        self.caH = calciumHline
    def redshift(self):
        z = ((self.caH - self.caH_0)/self.caH_0)
        return z
    def velocity(self):
        v = self.c*self.redshift()
        return v

gal2 = Galaxy("NGC1832", 3994.7)
print(f"Velocity of {gal2.name} is {gal2.velocity()}")
```

```python
import numpy as np
from matplotlib.pyplot import figure, show
from matplotlib.patches import Ellipse
import matplotlib.pyplot as plt
Nx = 400
Ny = 300
data = np.random.random((Ny,Nx))

fig=figure()
ax=fig.add_subplot(111)
ax.imshow(data, interpolation='none', origin="lower", cmap='gray')

x0=Nx/2
y0=Ny/2

a=100
b=50
angle=30.0

ellipse=Ellipse((x0,y0), width=2*a, height=2*b, angle=angle, edgecolor='red', facecolor='none')

ax.add_patch(ellipse)

plt.show()
```
